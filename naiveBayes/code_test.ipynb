# Cálculo de probabilidad con base en tablas de probabilidad de z-score

```python
def gaussian_pdf(x, mean, sd):
    return (1/(sd * ((2 * math.pi)**0.5))) * (math.e **(-(((x - mean) ** 2)/(2 * (sd ** 2)))))

def z_score(x, mean, sd):
    return (x - mean) / sd
```

```python
dataset_name = '../datasets/iris_mixed.csv'
norm_table_name = '../datasets/tablnorm.csv'

df = pd.read_csv(dataset_name, header=[0,1,2])
norm_table = pd.read_csv(norm_table_name, index_col=0)


metadata = df.columns
column_names = [valor[0].strip() for valor in metadata]
domine = [valor[1].strip() for valor in metadata]
data_type = [valor[2].strip() for valor in metadata]

df.columns = column_names

continuous_indexes = [idx for idx, value in enumerate(data_type) if value == 'numeric']

for index in continuous_indexes:
    mean = df.iloc[:, index].mean()
    sd = df.iloc[:, index].std()
    column_name = column_names[index]
    new_values = []
    for x in df.iloc[:, index]:
        new_values.append(z_score(x, mean, sd))
    df[column_name] = new_values

norm_table.loc[-4.0, '0']
```
